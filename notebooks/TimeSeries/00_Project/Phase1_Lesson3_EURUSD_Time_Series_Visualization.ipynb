## 1. Import libraries

````python
import pandas as pd
import matplotlib.pyplot as plt
````

## 2. Load the EUR/USD H4 dataset

The file path is:

`Data/EURUSD_H4.csv`


````python
eurusd = pd.read_csv(
    "Data/EURUSD_H4.csv",
    parse_dates=["DateTime"],
    index_col="DateTime"
)

eurusd = eurusd.sort_index()

eurusd.head()
````

## 3. Inspect the dataset

````python
eurusd.info()
````

````python
print(f"First timestamp: {eurusd.index.min()}")
print(f"Last timestamp : {eurusd.index.max()}")
print(f"Rows           : {len(eurusd):,}")
print(f"Columns        : {list(eurusd.columns)}")
````

## 4. Plot the full EUR/USD Close series

A line chart is usually the first visualization to try for time-series data because time order matters.


````python
plt.figure(figsize=(15, 6))

plt.plot(
    eurusd.index,
    eurusd["Close"]
)

plt.xlabel("Date")
plt.ylabel("EUR/USD")
plt.title("EUR/USD H4 Closing Price — Full Dataset")
plt.grid()

plt.show()
````

### What should you look for?

With the full dataset, focus on the big picture:

- broad upward or downward movements,
- major turning points,
- long periods of relatively high or low prices,
- periods with stronger or weaker fluctuations.

A long time range gives context, but it hides short-term detail.


## 5. Plot one year — 2025

````python
eurusd_2025 = eurusd.loc["2025"]

plt.figure(figsize=(15, 6))

plt.plot(
    eurusd_2025.index,
    eurusd_2025["Close"]
)

plt.xlabel("Date")
plt.ylabel("EUR/USD")
plt.title("EUR/USD H4 Closing Price — 2025")
plt.xticks(rotation=45)
plt.grid()

plt.show()
````

`plt.xticks(rotation=45)` rotates the x-axis date labels by 45 degrees, making them easier to read when dates are crowded.


## 6. Plot one month — November 2025

````python
eurusd_nov_2025 = eurusd.loc["2025-11"]

plt.figure(figsize=(15, 6))

plt.plot(
    eurusd_nov_2025.index,
    eurusd_nov_2025["Close"],
    marker="o"
)

plt.xlabel("Date")
plt.ylabel("EUR/USD")
plt.title("EUR/USD H4 Closing Price — November 2025")
plt.xticks(rotation=45)
plt.grid()

plt.show()
````

Markers are useful on shorter time windows because they make individual H4 observations easier to see. On very long ranges they can make a plot too crowded.


## 7. Plot one trading day

````python
eurusd_day = eurusd.loc["2025-11-05"]

plt.figure(figsize=(10, 5))

plt.plot(
    eurusd_day.index,
    eurusd_day["Close"],
    marker="o"
)

plt.xlabel("Time")
plt.ylabel("EUR/USD")
plt.title("EUR/USD H4 Closing Price — 2025-11-05")
plt.xticks(rotation=45)
plt.grid()

plt.show()
````

````python
print(f"Candles in this day: {len(eurusd_day)}")
print(f"Candle hours: {sorted(eurusd_day.index.hour.unique())}")
````

For H4 data, a complete trading day may contain candles at:

`00:00, 04:00, 08:00, 12:00, 16:00, 20:00`


## 8. Compare Open and Close

````python
plt.figure(figsize=(15, 6))

plt.plot(
    eurusd_nov_2025.index,
    eurusd_nov_2025["Open"],
    label="Open"
)

plt.plot(
    eurusd_nov_2025.index,
    eurusd_nov_2025["Close"],
    label="Close"
)

plt.xlabel("Date")
plt.ylabel("EUR/USD")
plt.title("EUR/USD H4 Open vs Close — November 2025")
plt.xticks(rotation=45)
plt.legend()
plt.grid()

plt.show()
````

When more than one line is plotted, `plt.legend()` identifies each series.


## 9. Compare High and Low

````python
plt.figure(figsize=(15, 6))

plt.plot(
    eurusd_nov_2025.index,
    eurusd_nov_2025["High"],
    label="High"
)

plt.plot(
    eurusd_nov_2025.index,
    eurusd_nov_2025["Low"],
    label="Low"
)

plt.xlabel("Date")
plt.ylabel("EUR/USD")
plt.title("EUR/USD H4 High vs Low — November 2025")
plt.xticks(rotation=45)
plt.legend()
plt.grid()

plt.show()
````

## 10. Plot all OHLC prices together

````python
plt.figure(figsize=(15, 6))

plt.plot(eurusd_nov_2025.index, eurusd_nov_2025["Open"], label="Open")
plt.plot(eurusd_nov_2025.index, eurusd_nov_2025["High"], label="High")
plt.plot(eurusd_nov_2025.index, eurusd_nov_2025["Low"], label="Low")
plt.plot(eurusd_nov_2025.index, eurusd_nov_2025["Close"], label="Close")

plt.xlabel("Date")
plt.ylabel("EUR/USD")
plt.title("EUR/USD H4 OHLC Prices — November 2025")
plt.xticks(rotation=45)
plt.legend()
plt.grid()

plt.show()
````

### Visualization lesson

More lines do not always make a better chart. If the chart becomes difficult to read, reduce the number of variables, shorten the time range, or both.


## 11. Plot Volume separately

````python
plt.figure(figsize=(15, 5))

plt.plot(
    eurusd_nov_2025.index,
    eurusd_nov_2025["Volume"]
)

plt.xlabel("Date")
plt.ylabel("Volume")
plt.title("EUR/USD H4 Volume — November 2025")
plt.xticks(rotation=45)
plt.grid()

plt.show()
````

## 12. Compare different time windows

The same time series can look very different depending on the time window.

- Full dataset → big-picture view
- One year → medium-term view
- One month → detailed view
- One day → individual H4 observations

Always choose the time window based on the question you are trying to answer.


## 13. Reusable plotting pattern

````python
data = eurusd.loc["2025-11"]

plt.figure(figsize=(15, 6))

plt.plot(
    data.index,
    data["Close"]
)

plt.xlabel("Date")
plt.ylabel("EUR/USD")
plt.title("EUR/USD Closing Price")
plt.xticks(rotation=45)
plt.grid()

plt.show()
````
