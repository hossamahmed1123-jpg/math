# Japanese Candlestick Chart Reference — EUR/USD H4

A compact reference notebook for plotting **Japanese candlestick charts** with `mplfinance`.

### Dataset
`Data/EURUSD_H4.csv`

### Expected columns
- Open
- High
- Low
- Close
- Volume

### Expected index
`DateTime`

## 2. Import libraries

````python
import pandas as pd
import mplfinance as mpf
````

## 3. Load the EUR/USD H4 dataset

````python
eurusd = pd.read_csv(
    "Data/EURUSD_H4.csv",
    parse_dates=["DateTime"],
    index_col="DateTime"
)

eurusd = eurusd.sort_index()

eurusd.head()
````

## 4. Basic Japanese candlestick chart

For candlesticks, a shorter time range is usually easier to read.


````python
eurusd_week = eurusd.loc["2025-11-03":"2025-11-07"]

mpf.plot(
    eurusd_week,
    type="candle",
    style="yahoo",
    title="EUR/USD H4 — Japanese Candlestick Chart",
    ylabel="EUR/USD",
    figsize=(15, 7)
)
````

## 5. Candlestick chart with Volume

````python
mpf.plot(
    eurusd_week,
    type="candle",
    style="yahoo",
    title="EUR/USD H4 — Japanese Candlestick Chart",
    ylabel="EUR/USD",
    volume=True,
    ylabel_lower="Volume",
    figsize=(15, 8)
)
````

## 6. Plot one month

````python
eurusd_month = eurusd.loc["2025-11"]

mpf.plot(
    eurusd_month,
    type="candle",
    style="yahoo",
    title="EUR/USD H4 — November 2025",
    ylabel="EUR/USD",
    volume=True,
    figsize=(16, 8)
)
````

## 7. Plot a custom date range

Change only `start_date` and `end_date`.


````python
start_date = "2025-11-10"
end_date = "2025-11-14"

chart_data = eurusd.loc[start_date:end_date]

mpf.plot(
    chart_data,
    type="candle",
    style="yahoo",
    title=f"EUR/USD H4 — {start_date} to {end_date}",
    ylabel="EUR/USD",
    volume=True,
    figsize=(15, 8)
)
````

## 8. Useful built-in styles

Try:

`yahoo`, `charles`, `binance`, `classic`, `default`, `nightclouds`


````python
mpf.plot(
    eurusd_week,
    type="candle",
    style="charles",
    title="EUR/USD H4 — Candlestick Chart",
    ylabel="EUR/USD",
    volume=True,
    figsize=(15, 8)
)
````

## 9. Show non-trading periods

Use `show_nontrading=True` if you want the x-axis to preserve missing/non-trading gaps.


````python
mpf.plot(
    eurusd_week,
    type="candle",
    style="yahoo",
    title="EUR/USD H4 — Showing Non-Trading Periods",
    ylabel="EUR/USD",
    volume=True,
    show_nontrading=True,
    figsize=(15, 8)
)
````

## 10. Quick reusable template

````python
start_date = "2025-11-03"
end_date = "2025-11-07"

data = eurusd.loc[start_date:end_date]

mpf.plot(
    data,
    type="candle",
    style="yahoo",
    title="EUR/USD H4",
    ylabel="EUR/USD",
    volume=True,
    figsize=(15, 8)
)
````

# Japanese Candlestick Cheat Sheet

Each H4 candle uses:

- **Open** → first price
- **High** → highest price
- **Low** → lowest price
- **Close** → last price

### Basic syntax

```python
mpf.plot(
    data,
    type="candle",
    style="yahoo",
    volume=True
)
```

### Useful options

```python
title="EUR/USD H4"
ylabel="EUR/USD"
volume=True
figsize=(15, 8)
show_nontrading=True
```

For H4 data, one week or one month is usually easier to read than the full dataset.
