```python
import pandas as pd

columns = [
    "Date",
    "Time",
    "Open",
    "High",
    "Low",
    "Close",
    "Volume"
]

eurusd = pd.read_csv(
    "DATA/EURUSD240.csv",
    header=None,
    names=columns
)

eurusd["DateTime"] = pd.to_datetime(
    eurusd["Date"] + " " + eurusd["Time"],
    format="%Y.%m.%d %H:%M"
)

eurusd = eurusd.set_index("DateTime")

eurusd = eurusd.drop(columns=["Date", "Time"])

eurusd = eurusd.sort_index()

eurusd.head()
```

```python
eurusd.index.to_series().diff().value_counts().head(10)
```

```python
print(pd.infer_freq(eurusd.index))
```

```python
eurusd.index.to_series().diff().mode()
```

```python
print("First observation:", eurusd.index.min())
print("Last observation :", eurusd.index.max())
print("Number of candles:", len(eurusd))
print("Base interval:", eurusd.index.to_series().diff().mode()[0])
```
