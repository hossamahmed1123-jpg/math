## 1. What is Time Series data?

A **Time Series** is a collection of observations recorded **over time**.

The key idea is:

> Each observation is associated with a specific point in time.

Example:

| Date | Sales |
|---|---:|
| Jan 2026 | 10,000 |
| Feb 2026 | 12,000 |
| Mar 2026 | 11,500 |
| Apr 2026 | 14,000 |

Here:

- `Sales` is the variable we are studying.
- `Date` tells us **when** each value happened.

We can represent a Time Series as:

\[
y_t
\]

where:

- \(y\) = the variable being studied
- \(t\) = time
- \(y_t\) = the value of the variable at time \(t\)


## 2. Time Series vs normal tabular data

Consider a normal customer dataset:

| Customer | Age | Salary |
|---|---:|---:|
| Ahmed | 25 | 8000 |
| Sara | 31 | 12000 |
| Omar | 28 | 9500 |

If we rearrange the rows, the meaning usually does not change much because each row describes a different customer.

Now consider a Time Series:

| Month | Sales |
|---|---:|
| January | 10,000 |
| February | 12,000 |
| March | 15,000 |
| April | 18,000 |

The order matters:

**January → February → March → April**

If we shuffle the observations, we destroy the time structure.

### Key idea

In Time Series, **chronological order contains information**.


## 3. Why chronological order matters

Suppose EUR/USD closes at:

| Day | EUR/USD |
|---|---:|
| Monday | 1.1700 |
| Tuesday | 1.1720 |
| Wednesday | 1.1750 |
| Thursday | 1.1730 |

When forecasting Thursday, we may use information from:

- Monday
- Tuesday
- Wednesday

But we must not use Friday because Friday has not happened yet.

This leads to a fundamental rule:

> **Future information must never be used to predict the past.**

Later, this becomes very important when studying:

- Train/test splitting
- Lag features
- Forecasting
- Backtesting
- Data leakage
- Look-ahead bias


## 4. What is a timestamp?

A **timestamp** tells us when an observation occurred.

Examples:

```text
2026-01-01
2026-01-02
2026-01-03
```

A timestamp can also include time:

```text
2026-01-01 10:00
2026-01-01 11:00
2026-01-01 12:00
```

A timestamp may represent:

- Year
- Month
- Day
- Hour
- Minute
- Second


## 5. What is frequency?

**Frequency** tells us how often observations are recorded.

Examples:

- `Jan 1 → Jan 2 → Jan 3` = **Daily**
- `10:00 → 11:00 → 12:00` = **Hourly**
- `Jan → Feb → Mar` = **Monthly**
- `2024 → 2025 → 2026` = **Yearly**

### Memory trick

> **Frequency = how much time passes between observations.**

In financial markets, common timeframes include:

- 1 minute
- 5 minutes
- 15 minutes
- 1 hour
- 4 hours
- 1 day
- 1 week


## 6. Why Time Series analysis is useful

Time Series analysis helps us answer questions such as:

- Is the series increasing or decreasing?
- Is there a repeating seasonal pattern?
- How volatile is the series?
- Does the past help explain the present?
- Can historical observations help forecast future observations?

Examples of Time Series data:

| Field | Example |
|---|---|
| Business | Monthly sales |
| Finance | Stock prices |
| Forex | EUR/USD prices |
| Economics | Inflation |
| Weather | Daily temperature |
| Energy | Hourly electricity demand |
| Web analytics | Daily website visitors |


## 7. Python example — Create a simple Time Series

We will create monthly sales data using Pandas.

The important steps are:

1. Create the DataFrame.
2. Convert the date column to a real datetime type.
3. Set the date as the index.


````python
import pandas as pd

data = {
    "Date": [
        "2026-01-01",
        "2026-02-01",
        "2026-03-01",
        "2026-04-01",
        "2026-05-01",
        "2026-06-01"
    ],
    "Sales": [
        10000,
        12000,
        11500,
        14000,
        15000,
        17000
    ]
}

df = pd.DataFrame(data)
df
````

### Convert the Date column to datetime

At first, the dates may be treated as ordinary text.

`pd.to_datetime()` converts them into a Pandas datetime type.


````python
df["Date"] = pd.to_datetime(df["Date"])

print(df.dtypes)
````

### Make time the index

For Time Series work, it is common to make the datetime column the DataFrame index.


````python
df = df.set_index("Date")

df
````

## 8. Visualize the Time Series

A **line chart** is one of the most important visualizations for Time Series because it preserves chronological order.

We will plot monthly sales over time.


````python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))
plt.plot(df.index, df["Sales"], marker="o")

plt.xlabel("Date")
plt.ylabel("Sales")
plt.title("Monthly Sales")

plt.show()
````

## 9. How to interpret the chart

The series generally moves upward.

There is a small decrease between February and March:

$$
12000 \rightarrow 11500
$$



After that, sales increase:

$$
11500 \rightarrow 14000 \rightarrow 15000 \rightarrow 17000
$$

This suggests a possible **upward trend**.

We will study trend formally in a later lesson.


## 10. Important forecasting idea

Suppose we want to predict June sales.

We may use:

- January
- February
- March
- April
- May

But we cannot use July, August, or September.

The forecasting structure is:

$$
\text{Past} \rightarrow \text{Present} \rightarrow \text{Future}
$$

Randomly shuffling a Time Series can allow future observations to appear in training data before earlier observations. This can create **data leakage** or **look-ahead bias**.
