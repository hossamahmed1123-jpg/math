# Phase 1 — Lesson 3: Time Series Visualization

In this notebook, you will practice plotting time-series data, resampling daily data to weekly data, and interpreting the chart.

## 1. Theory

When reading a time-series graph, look for:

- **Level**: the normal range of values.
- **Trend**: long-term increase or decrease.
- **Seasonality**: repeating pattern.
- **Noise**: random movement.
- **Unusual observations**: values that look very high or very low.

## 2. Import libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
```

## 3. Create the dataset

```python
data = {
    "Date": pd.date_range(start="2026-03-01", periods=30, freq="D"),
    "Visitors": [
        120, 125, 132, 128, 140,
        165, 180, 135, 138, 145,
        150, 160, 190, 205, 148,
        152, 158, 165, 170, 215,
        230, 160, 168, 172, 180,
        185, 220, 245, 190, 200
    ]
}

website = pd.DataFrame(data)
website = website.set_index("Date")
website.head()
```

## 4. Check the index and frequency

```python
type(website.index)
```

```python
website.index
```

## 5. Daily Website Visitors

```python
plt.figure(figsize=(10, 5))

plt.plot(
    website.index,
    website["Visitors"],
    marker="o"
)

plt.xlabel("Date")
plt.ylabel("Visitors")
plt.title("Daily Website Visitors")
plt.xticks(rotation=45)
plt.grid()
plt.tight_layout()

plt.show()
```

## 6. Weekly Average Visitors

```python
weekly = website[["Visitors"]].resample("W").mean()
weekly
```

```python
plt.figure(figsize=(10, 5))

plt.plot(
    weekly.index,
    weekly["Visitors"],
    marker="o"
)

plt.xlabel("Week")
plt.ylabel("Average Visitors")
plt.title("Average Weekly Website Visitors")
plt.grid()

plt.show()
```

## 7. Weekly Total Visitors

```python
weekly_total = website[["Visitors"]].resample("W").sum()
weekly_total
```

```python
plt.figure(figsize=(10, 5))

plt.plot(
    weekly_total.index,
    weekly_total["Visitors"],
    marker="o"
)

plt.xlabel("Week")
plt.ylabel("Total Visitors")
plt.title("Total Weekly Website Visitors")
plt.grid()

plt.show()
```

## 8. Daily vs Weekly Average

```python
plt.figure(figsize=(10, 5))

plt.plot(
    website.index,
    website["Visitors"],
    marker="o",
    label="Daily"
)

plt.plot(
    weekly.index,
    weekly["Visitors"],
    marker="o",
    label="Weekly Average"
)

plt.xlabel("Date")
plt.ylabel("Visitors")
plt.title("Daily vs Weekly Website Visitors")
plt.legend()
plt.grid()
plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

## 9. Maximum and Minimum Values

```python
highest_visitors = website["Visitors"].max()
highest_date = website["Visitors"].idxmax()

lowest_visitors = website["Visitors"].min()
lowest_date = website["Visitors"].idxmin()

print("Highest date:", highest_date)
print("Highest visitors:", highest_visitors)
print("Lowest date:", lowest_date)
print("Lowest visitors:", lowest_visitors)
```

## 10. Highlight the Maximum Point

```python
plt.figure(figsize=(10, 5))

plt.plot(
    website.index,
    website["Visitors"],
    marker="o"
)

plt.scatter(
    highest_date,
    highest_visitors,
    s=120
)

plt.annotate(
    "Highest Visitors",
    (highest_date, highest_visitors)
)

plt.xlabel("Date")
plt.ylabel("Visitors")
plt.title("Daily Website Visitors with Highest Point")
plt.xticks(rotation=45)
plt.grid()
plt.tight_layout()

plt.show()
```
