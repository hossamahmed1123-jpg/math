Now we study whether a dataset has one measured variable or multiple measured variables over time.

## 1. Univariate Time Series

A **univariate time series** contains one measured variable observed over time.

Important: the Date column is usually the **time index**, not a measured variable.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    "Date": pd.date_range(start="2026-01-01", periods=10, freq="D"),
    "Visitors": [100, 105, 110, 108, 115, 120, 125, 123, 130, 135]
}

df = pd.DataFrame(data)
df["Date"] = pd.to_datetime(df["Date"])
df = df.set_index("Date")
df
```

```python
plt.figure(figsize=(10, 5))
plt.plot(df.index, df["Visitors"], marker="o")
plt.xlabel("Date")
plt.ylabel("Visitors")
plt.title("Daily Visitors - Univariate Time Series")
plt.show()
```

## 2. Multivariate Time Series

A **multivariate time series** contains two or more measured variables observed over time.

```python
data = {
    "Date": pd.date_range(start="2026-04-01", periods=10, freq="D"),
    "Visitors": [100, 110, 115, 120, 125, 135, 140, 145, 150, 160],
    "Sales": [500, 520, 540, 570, 590, 630, 650, 680, 710, 750],
    "Ad_Spend": [100, 105, 110, 115, 120, 125, 130, 135, 140, 145]
}

multi_df = pd.DataFrame(data)
multi_df["Date"] = pd.to_datetime(multi_df["Date"])
multi_df = multi_df.set_index("Date")
multi_df
```

```python
multi_df.plot(figsize=(10, 5), marker="o")
plt.xlabel("Date")
plt.title("Multivariate Time Series")
plt.show()
```

## 3. Main Difference

| Type | Number of measured variables | Example |
|---|---:|---|
| Univariate | 1 | Sales |
| Multivariate | 2 or more | Sales + Visitors + Ad Spend |

Date is the time index. It is not counted as a measured variable.

## 4. Target and Explanatory Variables

If the goal is to forecast **Sales**, then **Sales** is the target variable.

Other variables such as **Visitors** and **Ad_Spend** may be explanatory or exogenous variables.

```python
# One column as a Series
sales_series = multi_df["Sales"]
type(sales_series)
```

```python
# Multiple columns as a DataFrame
sales_visitors_df = multi_df[["Sales", "Visitors"]]
type(sales_visitors_df)
```

```python
multi_df.corr()
```
